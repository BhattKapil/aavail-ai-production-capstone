# AAVAIL Capstone — Model Comparison
Compare forecasting models using validation MAE.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
r=pd.read_csv('../models/model_comparison.csv')
s=r.groupby('model',as_index=False).mae.mean()
s.plot.bar(x='model',y='mae',legend=False,title='Mean validation MAE'); plt.show()
```
