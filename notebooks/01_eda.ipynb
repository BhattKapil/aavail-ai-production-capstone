# AAVAIL Capstone — EDA
This notebook generates the EDA visualizations used in the capstone.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
df=pd.read_csv('../data/processed/daily_revenue.csv', parse_dates=['date'])
df.groupby('date').revenue.sum().plot(figsize=(12,5), title='Daily revenue'); plt.show()
df.pivot(index='date', columns='country', values='revenue').plot(figsize=(12,5), title='Revenue by country'); plt.show()
```
